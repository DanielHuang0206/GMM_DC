```python
# === 0. 套件 ===
import os, glob, time
import numpy as np, pandas as pd, cv2
import matplotlib.pyplot as plt
from typing import Dict, List
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
from pycpd import RigidRegistration  
from tqdm import tqdm  
from typing import Dict
# ---------------------------
# 1. 補點函式
# ---------------------------
#DTYPE = np.longdouble # or np.float64
import numpy as np
from sklearn.neighbors import NearestNeighbors

def ensure_two_points_in_circle_all(filtered_points, radius: float = 1.0):
    """
    對於 filtered_points 中的每個點 center：
      - 檢查以 center 為圓心、半徑 radius 的圓內，是否至少包含 2 個點（含 center 自身）。
      - 若不足，沿著「遠離最近鄰」的方向，於圓內新增 1 個點。
    回傳:
      updated_points: 原點集加上所有新點
      new_points:     本輪實際新增的點（若無則回傳 shape=(0,2) 的空陣列）
    介面與你的原函式一致。
    """
    pts = np.asarray(filtered_points, dtype=float)
    n = len(pts)
    if n < 1:
        # 與原行為一致：回傳 (原輸入, [])
        return pts, []

    # 為了效能：一次建模（KD-Tree / Ball-Tree 由 sklearn 自選）
    # 1) 半徑鄰居：判斷每個中心在半徑內的鄰居個數
    # 2) 最近鄰 (k=2)：取出每個點的最近「另一個」點（用來決定方向）
    nn_rad = NearestNeighbors(radius=radius, algorithm='auto').fit(pts)
    nn_k2  = NearestNeighbors(n_neighbors=min(2, n), algorithm='auto').fit(pts)

    # radius_neighbors：回傳每個點在半徑內的索引清單（含自身）
    rad_indices = nn_rad.radius_neighbors(pts, radius=radius, return_distance=False)

    # kneighbors：最近的 k 個點（含自身）
    d_k2, i_k2 = nn_k2.kneighbors(pts, return_distance=True)

    rng = np.random.default_rng()  # 可重現性若需要可改為帶 seed 的 Generator
    new_pts = []
    eps_dedup = radius * 1e-3       # 去重閾值
    max_try   = 8                   # 避免重疊時的最多嘗試次數
    place_r   = radius * 0.8        # 放置距離：留點餘量確保在圓內

    # 快速距離測試使用平方距離
    def too_close(p, arr, tol):
        if len(arr) == 0:
            return False
        d2 = np.sum((arr - p) ** 2, axis=1)
        return np.min(d2) < (tol * tol)

    for i, center in enumerate(pts):
        in_idx = rad_indices[i]
        if len(in_idx) >= 2:
            continue  # 這個中心的圓內已經有至少 2 個點

        # 只有自己一個點的情況
        if n == 1:
            # 隨機方向（單位向量）
            dir_vec = rng.normal(size=2)
            dir_norm = np.linalg.norm(dir_vec)
            if dir_norm < 1e-9:
                dir_vec = np.array([1.0, 0.0])
            else:
                dir_vec /= dir_norm

            candidate = center[:2] + dir_vec * place_r
            cand_full = candidate if center.shape[0] == 2 else np.concatenate([candidate, center[2:]])
            # 防重疊：嘗試旋轉方向
            tries = 0
            while too_close(cand_full[:2], pts[:, :2], eps_dedup) and tries < max_try:
                theta = (tries + 1) * (np.pi / 4.0)
                rot = np.array([[np.cos(theta), -np.sin(theta)],
                                [np.sin(theta),  np.cos(theta)]])
                cand2 = center[:2] + rot @ (dir_vec * place_r)
                cand_full = cand2 if center.shape[0] == 2 else np.concatenate([cand2, center[2:]])
                tries += 1

            new_pts.append(cand_full)
            continue

        # 多於 1 個點：往「遠離最近鄰」方向放一點
        # i_k2[i, 0] 是自己；若 n >= 2，i_k2[i, 1] 為最近的其他點
        nearest_idx = i_k2[i, 1] if n >= 2 else i
        nearest_point = pts[nearest_idx]

        # 方向：從最近鄰指向 center（遠離最近鄰）
        dir_vec = center[:2] - nearest_point[:2]
        dir_norm = np.linalg.norm(dir_vec)
        if dir_norm < 1e-9:
            # 假如重合，換隨機方向
            dir_vec = rng.normal(size=2)
            dir_norm = np.linalg.norm(dir_vec)
        dir_unit = dir_vec / (dir_norm + 1e-12)

        candidate = center[:2] + dir_unit * place_r
        cand_full = candidate if center.shape[0] == 2 else np.concatenate([candidate, center[2:]])

        # 防重疊：若太近則嘗試旋轉方向
        tries = 0
        while too_close(cand_full[:2], pts[:, :2], eps_dedup) and tries < max_try:
            theta = (tries + 1) * (np.pi / 6.0)  # 每次旋轉 30 度
            rot = np.array([[np.cos(theta), -np.sin(theta)],
                            [np.sin(theta),  np.cos(theta)]])
            cand2 = center[:2] + (rot @ dir_unit) * place_r
            cand_full = cand2 if center.shape[0] == 2 else np.concatenate([cand2, center[2:]])
            tries += 1

        new_pts.append(cand_full)

    if len(new_pts) > 0:
        new_pts = np.asarray(new_pts, dtype=float)
        updated_points = np.vstack([pts, new_pts])
    else:
        new_pts = np.empty((0, pts.shape[1]), dtype=float)
        updated_points = pts

    return updated_points, new_pts


def ensure_two_points_in_circle_iterative(points, radius: float = 1.0, max_iter: int = 1000):
    updated_points = points.copy()
    for _ in range(max_iter):
        updated_points, new_points = ensure_two_points_in_circle_all(updated_points, radius=radius)
        if len(new_points) == 0:
            break
    return updated_points
# ---------------------------
# 2. 雜訊過濾與輔助函式
# ---------------------------
def filter_noise_points_weighted(points, eps: float = 3, min_samples: int = 2,
                                 min_cluster_size: int = 55, return_mask: bool = False):
    if len(points) == 0:
        return np.array([]), np.array([]) if not return_mask else (np.array([]), np.array([]), None)
    db = DBSCAN(eps=eps, min_samples=min_samples).fit(points)
    labels = db.labels_
    unique_labels, counts = np.unique(labels[labels != -1], return_counts=True)
    large_clusters = set(unique_labels[counts >= min_cluster_size])
    mask = np.isin(labels, list(large_clusters))
    filtered_points = points[mask]
    return (filtered_points, mask) if not return_mask else (filtered_points, mask, mask)

def compute_local_density(points, k: int = 18):
    if len(points) == 0:
        return np.array([])
    actual_k = min(k + 1, len(points))
    nbrs = NearestNeighbors(n_neighbors=actual_k).fit(points)
    distances, _ = nbrs.kneighbors(points)
    avg_distance = np.mean(distances[:, 1:], axis=1)
    density = 1.0 / (avg_distance + 1e-6)
    return density

def estimate_rigid_transform(src, dst):
    src = np.asarray(src, dtype=np.float128, order='C', copy=False)
    dst = np.asarray(dst, dtype=np.float128, order='C', copy=False)
    centroid_src = np.mean(src, axis=0)
    centroid_dst = np.mean(dst, axis=0)
    src_centered = src - centroid_src
    dst_centered = dst - centroid_dst
    H = src_centered.T @ dst_centered
    U, _, Vt = np.linalg.svd(H)
    R = Vt.T @ U.T
    if np.linalg.det(R) < 0:
        Vt[1, :] *= -1
        R = Vt.T @ U.T
    t = centroid_dst - R @ centroid_src
    return R, t
def compute_point_weights(points, k=10):
    """根據局部密度計算權重。"""
    if len(points) <= k:
        return np.ones(len(points)) / len(points)
    nbrs = NearestNeighbors(n_neighbors=k+1).fit(points)
    distances, _ = nbrs.kneighbors(points)
    # 使用到第k個鄰居的平均距離的倒數作為密度代理
    avg_dist = np.mean(distances[:, 1:], axis=1)
    weights = 1.0 / (avg_dist + 1e-6)
    return weights / np.sum(weights) # 標準化權重
def pca_coarse_alignment(src: np.ndarray, dst: np.ndarray):

    if len(src) < 2 or len(dst) < 2:
        return np.eye(2), np.zeros(2)

    # 1. 計算權重
    src_weights = compute_point_weights(src)
    dst_weights = compute_point_weights(dst)

    # 2. 計算加權質心
    src_cent = np.average(src, axis=0, weights=src_weights)
    dst_cent = np.average(dst, axis=0, weights=dst_weights)

    # 3. 中心化
    src_c = src - src_cent
    dst_c = dst - dst_cent

    # 4. 計算加權共變異數矩陣
    Cov_src = src_c.T @ (src_c * src_weights[:, np.newaxis])
    Cov_dst = dst_c.T @ (dst_c * dst_weights[:, np.newaxis])
    
    # 後續步驟與標準PCA相同...
    U_src, _, _ = np.linalg.svd(Cov_src)
    U_dst, _, _ = np.linalg.svd(Cov_dst)

    R = U_dst @ U_src.T
    if np.linalg.det(R) < 0:
        U_dst[:, 1] *= -1
        R = U_dst @ U_src.T
    
    t = dst_cent - R @ src_cent
    return R, t

def usac_rigid_transform(src_points, dst_points, *, num_iterations: int = 100,
                         inlier_threshold: float = 50.0, min_inliers: int = 60):
    best_inlier_count = 0
    best_R = best_t = best_inlier_mask = None
    N = src_points.shape[0]
    if N < 2:
        raise ValueError("來源點數不足")

    for _ in range(num_iterations):
        if N < 8:
            break
        indices = np.random.choice(N, 8, replace=False)
        src_sample = src_points[indices]
        dst_sample = dst_points[indices]
        if np.linalg.norm(src_sample[1] - src_sample[0]) < 1e-10:
            continue
        R_cand, t_cand = estimate_rigid_transform(src_sample, dst_sample)
        src_transformed = (R_cand @ src_points.T).T + t_cand
        distances = np.linalg.norm(dst_points - src_transformed, axis=1)
        inlier_mask = distances < inlier_threshold
        inlier_count = np.sum(inlier_mask)
        if inlier_count > best_inlier_count and inlier_count >= min_inliers:
            best_inlier_count = inlier_count
            best_R, best_t, best_inlier_mask = R_cand, t_cand, inlier_mask
            break
    if best_inlier_mask is None or np.sum(best_inlier_mask) < min_inliers:
        raise ValueError("USAC 未找到足夠內點")
    src_inliers = src_points[best_inlier_mask]
    dst_inliers = dst_points[best_inlier_mask]
    best_R, best_t = estimate_rigid_transform(src_inliers, dst_inliers)
    return best_R, best_t, best_inlier_mask

# ---------------------------
# 3. ICP / CPD / Normal‑ICP
# ---------------------------

def icp_with_weights(source_points, target_points, weights, *, max_iterations: int = 5000, tolerance: float = 1e-20):#-20
    R_total = np.eye(2)
    t_total = np.zeros(2)
    prev_error = float("inf")
    for iteration in range(max_iterations):
        transformed_source = (R_total @ source_points.T).T + t_total
        nbrs = NearestNeighbors(n_neighbors=1).fit(target_points)
        distances, indices = nbrs.kneighbors(transformed_source)
        closest_points = target_points[indices.flatten()]
        error = np.sum(weights * (distances.flatten() ** 2)) / (np.sum(weights) + 1e-9)
        if abs(prev_error - error) < tolerance and error < 0.1:
            break
        prev_error = error
        src_centroid = np.average(transformed_source, axis=0, weights=weights)
        tgt_centroid = np.average(closest_points, axis=0, weights=weights)
        src_cent = transformed_source - src_centroid
        tgt_cent = closest_points - tgt_centroid
        H = (weights[:, None] * src_cent).T @ tgt_cent
        U, _, Vt = np.linalg.svd(H)
        R_delta = Vt.T @ U.T
        if np.linalg.det(R_delta) < 0:
            Vt[1, :] *= -1
            R_delta = Vt.T @ U.T
        t_delta = tgt_centroid - R_delta @ src_centroid
        R_total = R_delta @ R_total
        t_total = R_delta @ t_total + t_delta
    itr = iteration
    return R_total, t_total, prev_error, itr

def cpd_rigid_registration(source_points, target_points, *, max_iterations: int = 500, tolerance: float = 1e-3):
    reg = RigidRegistration(X=target_points, Y=source_points,
                            max_iterations=max_iterations, tolerance=tolerance)
    Y_reg, (s, R, t) = reg.register()
    return s, R, t, Y_reg

def compute_normals(points, k: int = 40):
    if len(points) < 3:
        return np.zeros_like(points)
    actual_k = min(k, len(points) - 1)
    nbrs = NearestNeighbors(n_neighbors=actual_k).fit(points)
    _, indices = nbrs.kneighbors(points)
    normals = np.zeros_like(points)
    for i, neighbors in enumerate(indices):
        cov_matrix = np.cov(points[neighbors].T)
        eigvals, eigvecs = np.linalg.eigh(cov_matrix)
        normals[i] = eigvecs[:, 0]
    return normals / (np.linalg.norm(normals, axis=1, keepdims=True) + 1e-9)
"""
def normal_icp(source_points, target_points, weights, *, alpha: float = 0.1, gamma: float = 0.8,
               max_iterations: int = 5000, tolerance: float = 1e-20):
    source_normals = compute_normals(source_points)
    target_normals = compute_normals(target_points)
    R_total = np.eye(2)
    t_total = np.zeros(2)
    prev_error = float("inf")
    for iteration in range(max_iterations):
        #print(iteration)
        transformed_source = (R_total @ source_points.T).T + t_total
        nbrs = NearestNeighbors(n_neighbors=1).fit(target_points)
        distances, indices = nbrs.kneighbors(transformed_source)
        indices = np.clip(indices.flatten(), 0, len(target_normals) - 1)
        closest_points = target_points[indices]
        closest_normals = target_normals[indices]
        normal_dot = np.einsum("ij,ij->i", source_normals, closest_normals)
        normal_errors = np.arccos(np.clip(normal_dot, -1, 1))
        weighted_normals = np.exp(-normal_errors / (2 * alpha)) ** gamma
        combined_weights = weights * distances.flatten() * weighted_normals
        error = np.sum(combined_weights * (distances.flatten() ** 2)) / (np.sum(combined_weights) + 1e-9)
        if abs(prev_error - error) < tolerance and error < 0.1:
            break
        prev_error = error
        src_centroid = np.average(transformed_source, axis=0, weights=combined_weights)
        tgt_centroid = np.average(closest_points, axis=0, weights=combined_weights)
        src_cent = transformed_source - src_centroid
        tgt_cent = closest_points - tgt_centroid
        H = (combined_weights[:, None] * src_cent).T @ tgt_cent
        U, _, Vt = np.linalg.svd(H)
        R_delta = Vt.T @ U.T
        if np.linalg.det(R_delta) < 0:
            Vt[1, :] *= -1
            R_delta = Vt.T @ U.T
        t_delta = tgt_centroid - R_delta @ src_centroid
        R_total = R_delta @ R_total
        t_total = R_delta @ t_total + t_delta
    itr = iteration
    return R_total, t_total, prev_error,itr
"""
def normal_icp(source_points, target_points, weights, *, alpha: float = 0.09, gamma: float = 0.8,
               max_iterations: int = 5000, tolerance: float = 1e-20):#0.08 0.8 5000 -20
    source_normals = compute_normals(source_points)
    target_normals = compute_normals(target_points)
    R_total = np.eye(2)
    t_total = np.zeros(2)
    prev_error = float("inf")
    for iteration in range(max_iterations):
        #print(iteration)
        transformed_source = (R_total @ source_points.T).T + t_total
        nbrs = NearestNeighbors(n_neighbors=1).fit(target_points)
        distances, indices = nbrs.kneighbors(transformed_source)
        indices = np.clip(indices.flatten(), 0, len(target_normals) - 1)
        closest_points = target_points[indices]
        closest_normals = target_normals[indices]

        # === 修正部分 START ===
        
        # 1. 將來源法向量用當前的 R_total 進行旋轉
        transformed_normals = (R_total @ source_normals.T).T
        
        # 2. 使用旋轉後的法向量來計算點積
        normal_dot = np.einsum("ij,ij->i", transformed_normals, closest_normals)

        # === 修正部分 END ===

        normal_errors = np.arccos(np.clip(normal_dot, -1, 1))
        weighted_normals = np.exp(-normal_errors / (2 * alpha)) ** gamma
        combined_weights = weights * distances.flatten() * weighted_normals
        error = np.sum(combined_weights * (distances.flatten() ** 2)) / (np.sum(combined_weights) + 1e-9)
        if abs(prev_error - error) < tolerance and error < 0.3:
            break
        prev_error = error
        src_centroid = np.average(transformed_source, axis=0, weights=combined_weights)
        tgt_centroid = np.average(closest_points, axis=0, weights=combined_weights)
        src_cent = transformed_source - src_centroid
        tgt_cent = closest_points - tgt_centroid
        H = (combined_weights[:, None] * src_cent).T @ tgt_cent
        U, _, Vt = np.linalg.svd(H)
        R_delta = Vt.T @ U.T
        if np.linalg.det(R_delta) < 0:
            Vt[1, :] *= -1
            R_delta = Vt.T @ U.T
        t_delta = tgt_centroid - R_delta @ src_centroid
        R_total = R_delta @ R_total
        t_total = R_delta @ t_total + t_delta
    itr = iteration
    return R_total, t_total, prev_error,itr
"""
def fill_line_gaps(points, *, max_distance: float = 0.05, max_iter: int = 5):#3 5
    pts = points.copy()
    for _ in range(max_iter):
        nbrs = NearestNeighbors(n_neighbors=2).fit(pts)
        distances, indices = nbrs.kneighbors(pts)
        new_pts = []
        for i in range(len(pts)):
            d = distances[i, 1]
            if d > max_distance:
                p1, p2 = pts[i], pts[indices[i, 1]]
                num_insert = int(d // max_distance)
                for n in range(1, num_insert + 1):
                    new_pts.append(p1 + (p2 - p1) * n / (num_insert + 1))
        if not new_pts:
            break
        pts = np.vstack([pts, np.array(new_pts)])
    return pts
"""
import numpy as np
from sklearn.neighbors import NearestNeighbors

def fill_line_gaps(points, *, base_factor: float = 0.05, k_neighbors: int = 8, max_iter: int = 5000):
    """
    自適應補點：
      - 對每個點計算最近 k 個鄰居的平均距離，作為該點的局部基準距離 local_d。
      - 若某條邊長 dist > base_factor * local_d，則在中間插值。
      - 插值數量 ≈ dist / local_d。
    參數:
      base_factor: 控制補點敏感度 (越小越容易補)
      k_neighbors: 用多少個鄰居估算局部密度
      max_iter: 最多迭代次數
    輸出:
      pts: (N+新增, D) 的 numpy.ndarray
    """
    pts = np.asarray(points).copy()
    if pts.ndim != 2 or len(pts) < 2:
        return pts

    eps_dedup = 1e-9  # 避免重複點

    def dedup_append(base_pts, new_pts, radius):
        if len(new_pts) == 0:
            return base_pts
        keep = []
        for q in new_pts:
            d2 = np.sum((base_pts - q) ** 2, axis=1)
            if d2.min() >= radius * radius:
                keep.append(q)
        if keep:
            return np.vstack([base_pts, np.vstack(keep)])
        return base_pts

    n_iter = 0
    while True:
        if (max_iter is not None) and (n_iter >= max_iter):
            break
        n_iter += 1

        # 最近鄰模型
        nbrs = NearestNeighbors(n_neighbors=min(k_neighbors+1, len(pts)), algorithm='auto').fit(pts)
        distances, indices = nbrs.kneighbors(pts, return_distance=True)

        # 局部平均距離 (排除自身)
        local_d = distances[:, 1:].mean(axis=1)

        # 蒐集需要補點的邊
        new_chunks = []
        for i in range(len(pts)):
            for j in indices[i, 1:2]:  # 只考慮最近的一個鄰居即可
                if j <= i: 
                    continue  # 避免重複邊
                dist = np.linalg.norm(pts[j] - pts[i])
                th = base_factor * 0.5 * (local_d[i] + local_d[j])  # 平均兩端局部距離
                if dist > th:
                    n_ins = int(np.floor(dist / th))
                    if n_ins > 0:
                        t = np.linspace(1, n_ins, n_ins) / (n_ins + 1)
                        seg = pts[i] + (pts[j] - pts[i]) * t[:, None]
                        new_chunks.append(seg)

        if not new_chunks:
            break

        new_pts = np.vstack(new_chunks)
        pts = dedup_append(pts, new_pts, eps_dedup)

    return pts



def pca_coarse_alignment(src: np.ndarray, dst: np.ndarray):
    """
    強化版PCA：修正了180度的方向模糊性。
    """
    # 1. 計算質心
    src_cent = src.mean(axis=0)
    dst_cent = dst.mean(axis=0)

    # 2. 中心化
    src_c = src - src_cent
    dst_c = dst - dst_cent

    # 3. 分別計算共變異數矩陣
    Cov_src = src_c.T @ src_c
    Cov_dst = dst_c.T @ dst_c

    # 4. SVD 分解得到主方向
    U_src, _, _ = np.linalg.svd(Cov_src)
    U_dst, _, _ = np.linalg.svd(Cov_dst)

    # === 新增的強化步驟 ===
    # 檢查第一主軸方向，若相反則翻轉
    if np.dot(U_src[:, 0], U_dst[:, 0]) < 0:
        U_dst[:, 0] *= -1
    # 檢查第二主軸方向
    if np.dot(U_src[:, 1], U_dst[:, 1]) < 0:
        U_dst[:, 1] *= -1
    # ======================

    # 5. 主軸對齊
    R = U_dst @ U_src.T

    # 6. 處理鏡像情況 (理論上，修正方向後這一步的需求會降低)
    if np.linalg.det(R) < 0:
        U_dst[:, 1] *= -1
        R = U_dst @ U_src.T

    # 7. 平移對齊質心
    t = dst_cent - R @ src_cent

    return R, t
# ---------------------------
# 4. DATASETS 參數清單
# ---------------------------
"""
        "name": "red01",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red01",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01/Image_20240321100710525.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321100710525.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red01_combined.xlsx",
    },
    {
        "name": "red02",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red02",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red02",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red02/Image_20240321104026349.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321104026349.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red02_combined.xlsx",
    },
    {
        "name": "red03",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red03",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red03",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red03/Image_20240321105545542.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321105545542.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red03_combined.xlsx",
    },
    {
        "name": "red04",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red04",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red04",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red04/Image_20240321111055812.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321111055812.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red04_combined.xlsx",
    },
    {
        "name": "red05",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red05",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red05",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red05/Image_20240321112738655.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321112738655.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red05_combined.xlsx",
    },
    {
        "name": "red06",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red06",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red06",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red06/Image_20240321114408009.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321114408009.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red06_combined.xlsx",
    },
    {
        "name": "white01",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white01",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white01",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white01/Image_20240321143412883.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321143412883.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white01_combined.xlsx",
    },
    {
        "name": "white02",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white02",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white02",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white02/Image_20240321145312162.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321145312162.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white02_combined.xlsx",
    },
    {
        "name": "white03",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white03",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white03",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white03/Image_20240321151040617.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321151040617.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white03_combined.xlsx",
    },
    {
        "name": "white04",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white04",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white04",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white04/Image_20240321152853920.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321152853920.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white04_combined.xlsx",
    },
    {
        "name": "white05",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white05",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white05",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white05/Image_20240321154611642.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321154611642.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white05_combined.xlsx",
    },
    {
        "name": "white06",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white06",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white06",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white06/Image_20240321160945507.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321160945507.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white06_combined.xlsx",
    },
        "name": "red01",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red01",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01/Image_20240321100710525.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321100710525.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red01_combined.xlsx",
    },
        {

         "name": "white01",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white01",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/white01",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/white01/Image_20240321143412883.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321143412883.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/white01_combined.xlsx",
    },
"""
DATASETS: List[Dict] = [
    {

        "name": "red01",
        "csv_dir":  r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01",
        "img_dir":  r"/media/dc0206/Crucial X6/GMM20/20240329_DATA2/20240329_DATA/NTHU_5x/red01",
        "template_csv": r"/media/dc0206/Crucial X6/GMM20/EdgeFinderResult/red01/Image_20240321100710525.csv",
        "template_img": r"/media/dc0206/Crucial X6/GMM20/crop_w5.bmp",
        "first_img": "Image_20240321100710525.bmp",
        "final_xlsx": r"/media/dc0206/Crucial X6/GMM20/red01_combined.xlsx",
    },

    # --- 新資料集請直接在此處繼續加入 ------------------------------------
]
def randomly_block_points(points, min_block_size=5, max_block_size=15):
    """
    從點雲中隨機移除一個矩形區塊的點。

    Args:
        points (np.ndarray): 輸入的點雲，形狀為 (N, 2)，N是點的數量。
        min_block_size (float): 遮擋區塊的最小尺寸。
        max_block_size (float): 遮擋區塊的最大尺寸。

    Returns:
        np.ndarray: 移除了部分點之後的新點雲。
    """
    if len(points) == 0:
        return points

    # 1. 隨機選擇一個現有的點作為遮擋區塊的中心
    center_index = np.random.randint(0, len(points))
    center_point = points[center_index]
    center_x, center_y = center_point[0], center_point[1]

    # 2. 隨機決定遮擋區塊的寬和高
    block_width = np.random.uniform(min_block_size, max_block_size)
    block_height = np.random.uniform(min_block_size, max_block_size)

    # 計算區塊的邊界
    x_min = center_x - block_width / 2
    x_max = center_x + block_width / 2
    y_min = center_y - block_height / 2
    y_max = center_y + block_height / 2

    # 3. 找出所有不在區塊內的點 (使用 NumPy Boolean Masking)
    # 建立一個布林遮罩 (mask)，標記出在區塊內的點
    points_in_block_mask = (
        (points[:, 0] >= x_min) & (points[:, 0] <= x_max) &
        (points[:, 1] >= y_min) & (points[:, 1] <= y_max)
    )

    # 使用遮罩的反向 (~) 來選取所有"不在"區塊內的點
    return points[~points_in_block_mask]

# 5. 單資料集完整流程


# ──────────────────────────────────────────────────────────────────────────────
# 主要流程
# ──────────────────────────────────────────────────────────────────────────────
def process_dataset(cfg: Dict):
    """依 cfg 設定批次處理資料集，並輸出對齊結果與耗時統計。"""
    t_start = time.perf_counter()

    # === 0. 從 cfg 讀路徑與旗標（可自行在 cfg 補充） ===
    csv_dir        = cfg["csv_dir"]
    img_dir        = cfg["img_dir"]
    template_csv   = cfg["template_csv"]
    template_img_p = cfg["template_img"]
    first_img_name = cfg["first_img"]
    final_xlsx     = cfg["final_xlsx"]

    # 旗標（可在 cfg 覆寫）
    do_template   = cfg.get("do_template",   True)
    do_preprocess = cfg.get("do_preprocess", True)
    do_pca        = cfg.get("do_pca",        True)
    do_usac       = cfg.get("do_usac",       False)
    do_icp        = cfg.get("do_icp",        True)
    do_cpd        = cfg.get("do_cpd",        True)
    #抽樣
    sample = 1.0
    # === 1. 耗時記錄器 ===
    step_times = {k: [] for k in ["template", "csv", "denoise", "PCA" ,"usac", "fine"]}
    def log_time(dic, key, t0):
        dic[key].append(time.perf_counter() - t0)

    # === 2. 讀取模板並建立初始參考 ===
    t0 = time.perf_counter()
    tpl_df  = pd.read_csv(template_csv)
    tpl_pts = tpl_df[['PosX', 'PosY']].values
    tpl_img = cv2.imread(template_img_p, cv2.IMREAD_GRAYSCALE)
    first_im = cv2.imread(os.path.join(img_dir, first_img_name),
                          cv2.IMREAD_GRAYSCALE)
    res = cv2.matchTemplate(first_im, tpl_img, cv2.TM_CCOEFF_NORMED)
    _, _, _, first_tl = cv2.minMaxLoc(res)
    w_tpl, h_tpl = tpl_img.shape[1], tpl_img.shape[0]
    h,w = tpl_img.shape[:2]
    log_time(step_times, "template", t0)

    # 篩掉模板範圍外點
    first_br = (first_tl[0] + w_tpl, first_tl[1] + h_tpl)
    tpl_mask = (
        (tpl_pts[:, 0] >= first_tl[0]) & (tpl_pts[:, 0] <= first_br[0]) &
        (tpl_pts[:, 1] >= first_tl[1]) & (tpl_pts[:, 1] <= first_br[1])
    )
    tpl_pts = tpl_pts[tpl_mask]
    if len(tpl_pts) < 3:
        raise ValueError("模板區域點不足，請確認模板與 CSV 座標。")

    # 讀第一張 CSV 建立 ref
    t0 = time.perf_counter()
    df_first = pd.read_csv(os.path.join(csv_dir, first_img_name.replace(".bmp", ".csv")))

    # 以 first_tl / first_br 做 ROI 篩選
    first_mask = (
        (df_first['PosX'] >= first_tl[0]) & (df_first['PosX'] <= first_br[0]) &
        (df_first['PosY'] >= first_tl[1]) & (df_first['PosY'] <= first_br[1])
    )
    ref_df = df_first.loc[first_mask, ['PosX', 'PosY']]
    ref_df_numpy = ref_df.to_numpy()
    ref_df_numpy = filter_noise_points_weighted(ref_df_numpy)[0]
    import matplotlib.pyplot as plt

    # 安全抽樣 90%（可重現）
    if len(ref_df) > 0:
        ref = ref_df.sample(frac= sample, random_state=19).to_numpy()
    else:
        ref = np.empty((0, 2), dtype=float)


    ref = filter_noise_points_weighted(ref)[0]
    


    ref = ensure_two_points_in_circle_iterative(ref, radius=1.5, max_iter=3000)
    ref = ensure_two_points_in_circle_iterative(ref, radius=1.2, max_iter=3000)
    ref = ensure_two_points_in_circle_iterative(ref, radius=0.9, max_iter=3000)


    
    ref = fill_line_gaps(ref, base_factor=0.5, k_neighbors=8)
    """
    plt.scatter(ref[:, 0], ref[:, 1], alpha=1.0)
    plt.title("fill_line_gaps")
    plt.xlabel("Column 0")
    plt.ylabel("Column 1")
    plt.show()
    """
    log_time(step_times, "csv", t0)

    # === 3. 初始化結果表 ===
    results = [{
        "image_file": first_img_name,
        "rotation_matrix": np.eye(2).tolist(),
        "translation_vector": [0.0, 0.0],
        "incremental_displacement": [0.0, 0.0],
        "incremental_disp_norm": 0.0,
        "displacement": [0.0, 0.0],
        "displacement_norm": 0.0,
        "alignment_error": 0.0,
    }]
    prev_inc = np.zeros(2)

    # === 4. 逐張影像處理 ===
    imgs = sorted(glob.glob(os.path.join(img_dir, "*.bmp")))
    csvs = sorted(glob.glob(os.path.join(csv_dir, "*.csv")))

    # 使用 tqdm 追蹤進度
    for idx in tqdm(range(1, len(imgs)), desc="Processing images", unit="img"):
        img_p  = imgs[idx]
        csv_p  = csvs[idx]
        im     = cv2.imread(img_p, cv2.IMREAD_GRAYSCALE)
        if im is None:
            continue

        # 4-1 ROI（模板匹配）
        t0 = time.perf_counter()
        if do_template:
            res = cv2.matchTemplate(im, tpl_img, cv2.TM_CCOEFF_NORMED)
            _, _, _, tl = cv2.minMaxLoc(res)          # ROI 左上角
            roi_tl = (int(tl[0]), int(tl[1]))
            h, w = tpl_img.shape[:2]                   # ROI 尺寸 = 模板尺寸
            roi_br = (roi_tl[0] + w, roi_tl[1] + h)    # ROI 右下角
            region_center = np.array([roi_tl[0] + w/2.0, roi_tl[1] + h/2.0])
        else:
            H, W = im.shape[:2]
            roi_tl = (0, 0)
            roi_br = (W - 1, H - 1)
            region_center = np.array([W/2.0, H/2.0])   # 無模板時用整張圖中心
        log_time(step_times, "template", t0)

        # 4-2 讀 CSV + 篩 ROI + 抽樣 90% → 產生 cur
        t0 = time.perf_counter()
        dfc = pd.read_csv(csv_p)
        roi_mask = (
            (dfc['PosX'] >= roi_tl[0]) & (dfc['PosX'] <= roi_br[0]) &
            (dfc['PosY'] >= roi_tl[1]) & (dfc['PosY'] <= roi_br[1])
        )
        cur_df = dfc.loc[roi_mask, ['PosX', 'PosY']]
        if len(cur_df) > 0:
            cur = cur_df.sample(frac=sample , random_state=19).to_numpy()
        else:
            cur = np.empty((0, 2), dtype=float)
        log_time(step_times, "csv", t0)

        if len(cur) < 3 or len(ref) < 3:
            continue


        # 4-3 雜訊過濾 / 補點
        t0 = time.perf_counter()
        cur = filter_noise_points_weighted(cur)[0]
        cur = randomly_block_points(cur) 
        """
        plt.scatter(cur[:, 0], cur[:, 1], alpha=1.0)
        plt.title("Scatter before")
        plt.xlabel("Column 0")
        plt.ylabel("Column 1")
        plt.show()
        """
        if do_preprocess:
            cur = ensure_two_points_in_circle_iterative(cur, radius=1.5,
                                                        max_iter=3000)
            cur = ensure_two_points_in_circle_iterative(cur, radius=1.2,
                                                        max_iter=3000)
            cur = ensure_two_points_in_circle_iterative(cur, radius=0.9,
                                                        max_iter=3000)
            cur = fill_line_gaps(cur, base_factor=0.5, k_neighbors=8)
        log_time(step_times, "denoise", t0)
        """
        plt.scatter(cur[:, 0], cur[:, 1], alpha=1.0)
        plt.title("Scatter after")
        plt.xlabel("Column 0")
        plt.ylabel("Column 1")
        plt.show()
        """
        if do_pca:
            R_pca, t_pca  = pca_coarse_alignment(cur, ref)
        else:
            R_pca, t_pca  = np.eye(2), np.zeros(2)
        
        cur_u = (R_pca @ cur.T).T + t_pca
        log_time(step_times, "PCA", t0)
        # 4-4 USAC
        t0 = time.perf_counter()
        if do_usac:
            nbrs = NearestNeighbors(n_neighbors=1).fit(ref)
            _, idx2 = nbrs.kneighbors(cur)
            tgt = ref[idx2.flatten()]
            try:
                R_u, t_u, _ = usac_rigid_transform(
                    cur, tgt, inlier_threshold=40, min_inliers=20)
                
            except Exception:
                R_u, t_u = np.eye(2), np.zeros(2)
                
        else:
            R_u, t_u = np.eye(2), np.zeros(2)
            
        cur_u = (R_u @ cur.T).T + t_u
        log_time(step_times, "usac", t0)

        # 4-5 ICP / CPD / Normal-ICP
        t0 = time.perf_counter()
        R_f, t_f, err_f = np.eye(2), np.zeros(2), None
        dens = compute_local_density(cur_u, k=25)
        dens = (dens - dens.min()) / (dens.max() - dens.min() + 1e-9)

        if do_icp and do_cpd:
            #R_p, t_p, e_p = icp_with_weights(cur_u, ref, dens)
            R_p, t_p, e_p, itr = normal_icp(cur_u, ref, dens)
            
            try:
                _, R_c, t_c, _ = cpd_rigid_registration(
                    cur_u, ref, max_iterations=100, tolerance=1e-6)#100
            except Exception:
                R_c, t_c = np.eye(2), np.zeros(2)
            R_n, t_n, e_n, itr = normal_icp((R_c @ cur_u.T).T + t_c, ref, dens)
            if e_p < e_n:
                R_f, t_f, err_f = R_p, t_p, e_p
                print("ICP:")
                print(itr)
            else:
                R_f = R_n @ R_c
                t_f = R_n @ t_c + t_n
                err_f = e_n
                print("CPD")
                print(itr)
        elif do_icp:
            R_f, t_f, err_f, itr = icp_with_weights(cur_u, ref, dens)
            print("ICP:")
            print(itr)
        elif do_cpd:
            try:
                _, R_c, t_c, _ = cpd_rigid_registration(
                    cur_u, ref, max_iterations=100, tolerance=1e-3)#100
            except Exception:
                R_c, t_c = np.eye(2), np.zeros(2)
            R_n, t_n, err_f, itr = normal_icp((R_c @ cur_u.T).T + t_c, ref)
            R_f = R_n @ R_c
            t_f = R_n @ t_c + t_n
            print("ICP:")
            print(itr)
        log_time(step_times, "fine", t0)
        transformed_center = R_f @ region_center + t_f
        inc = transformed_center - region_center
        disp = inc - prev_inc
        prev_inc = inc.copy()
        cur_final = (R_f @ cur_u.T).T + t_f
        plt.scatter(ref_df_numpy[:, 0], ref_df_numpy[:, 1], alpha=1.0)
        plt.scatter(cur_final[:, 0], cur_final[:, 1], alpha=1.0)
        plt.title("fill_line_gaps")
        plt.xlabel("Column 0")
        plt.ylabel("Column 1")
        plt.show()
        # 2-8 收集結果
        results.append({
            "image_file": os.path.basename(imgs[idx]),
            "rotation_matrix":          R_f.tolist(),
            "translation_vector":       t_f.tolist(),
            "incremental_displacement": inc.tolist(),
            "incremental_disp_norm":    float(np.linalg.norm(inc)),
            "displacement":             disp.tolist(),
            "displacement_norm":        float(np.linalg.norm(disp)),
            "alignment_error":          float(err_f),
        })

    # 結束時可根據步驟統計資訊輸出耗時等......
    total_time = time.perf_counter() - t_start
    print(f"Total processing time: {total_time:.2f} seconds")


    # === 5. 輸出 Excel ===
    os.makedirs(os.path.dirname(final_xlsx), exist_ok=True)
    pd.DataFrame(results).to_excel(final_xlsx, index=False)

    # === 6. 畫圓餅圖 ===
    labels = list(step_times.keys())
    totals = [sum(step_times[k]) for k in labels]
    if sum(totals) > 0:
        plt.figure(figsize=(6, 6))
        plt.pie(totals, labels=labels, autopct="%1.1f%%", startangle=45)
        plt.axis("equal")
        chart_dir = os.path.join(os.path.dirname(final_xlsx), "time_charts")
        os.makedirs(chart_dir, exist_ok=True)
        plt.savefig(os.path.join(chart_dir, f"{cfg['name']}_time_pie.png"),
                    dpi=300, bbox_inches="tight")
        plt.close()

    # === 7. 完成訊息 ===
    elapsed = time.perf_counter() - t_start
    print(f"✅ [{cfg['name']}] 完成，輸出：{final_xlsx}，總耗時 {elapsed:.1f}s")


def main():
    for cfg in DATASETS:
        try:
            process_dataset(cfg)
        except Exception as e:
            print(f"[{cfg['name']}] 失敗：{e}")

if __name__ == "__main__":
    main()

```
